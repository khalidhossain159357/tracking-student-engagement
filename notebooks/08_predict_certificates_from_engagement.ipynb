## Importing Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import seaborn as sns

sns.set()
```

## Reading the CSV File

```python
raw_data = pd.read_csv('/kaggle/input/datasets/khalidhossain159357/student-engagement-info/minutes_and_certificates.csv')

data = raw_data.copy()

data.head()
```

## Splitting the Data into Inputs & Target

```python
target = data['certificates_issued']
inputs = data['minutes_watched']
```

## Splitting the Data into Training & Test Sets

```python
x_train, x_test, y_train, y_test = train_test_split(inputs, 
                                                    target, 
                                                    test_size=0.2, 
                                                    random_state=365)

x_train = x_train.to_numpy() 
x_test = x_test.to_numpy() 
y_train = y_train.to_numpy() 
y_test = y_test.to_numpy() 

x_train = x_train.reshape(-1, 1)
x_test = x_test.reshape(-1, 1)
```

## Creating the Model & Fitting it to the Data

```python
reg = LinearRegression()

reg.fit(x_train,y_train)
```

```python
reg.intercept_, reg.coef_
```

## Calculating the R-Squared Value 

```python
reg.score(x_train, y_train)
```

## Making Predictions

```python
y_hat_test = reg.predict(x_test)
```

```python
reg.predict([[1200]])
```

## Plotting the Results

```python
plt.scatter(y_test, y_hat_test)
plt.xlabel('Targets (y_test)',size=18)
plt.ylabel('Predictions (y_hat_test)',size=18)
plt.show()
```
